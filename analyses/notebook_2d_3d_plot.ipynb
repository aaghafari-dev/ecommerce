```python
# check the raw data whether can be loaded
df1 = pd.read_csv(Raw_data_path1)
df1.head(3)
```

```python
df2 = pd.read_csv(Raw_data_path2)
df2.head(3)
```

```python
# Prepare data for heatmap
heatmap_data = df1.pivot(
    index="CAL_YEAR",
    columns="CAL_MONTH",
    values="TOTAL_SALES"
)
```

```python
# Plot 2D heatmap 
plt.figure(figsize=(10, 6))

plt.imshow(heatmap_data, aspect="auto")
plt.colorbar(label="Total Sales")

plt.xticks(
    ticks=range(len(heatmap_data.columns)),
    labels=heatmap_data.columns
)
plt.yticks(
    ticks=range(len(heatmap_data.index)),
    labels=heatmap_data.index
)

plt.xlabel("Month")
plt.ylabel("Year")
plt.title("2D Heatmap of Sales by Year and Month")

plt.show()
```

```python
# Prepare data for 3D plotting
x = df2["CAL_MONTH"]
y = df2["CAL_YEAR"]
z = df2["TOTAL_SALES"]
```

```python
# Create 3D scatter plot
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection="3d")

ax.scatter(x, y, z)

ax.set_xlabel("Month")
ax.set_ylabel("Year")
ax.set_zlabel("Total Sales")

ax.set_title("3D Sales Analysis by Year and Month")

plt.show()
```
